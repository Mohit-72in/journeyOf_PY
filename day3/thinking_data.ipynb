```python
import json
```

```python
#lode the data
def load_data(filename):
    with open(filename,'r') as f:
        data = json.load(f)
    return data
```

```python
data = load_data("store_data.json")
print(data)
print(type(data))
```

```python
#cleaning and structuring our data
# 1: data is inconsistent : mixed type consistent/clean data value like ex here rating has int,float,and string type value stored
# 2: missing data : replace missing field with none age or handle missing value
# 3: duplicate data : deduplication means removing duplicate data having unique value only
```

```python
def clean_data(data):
    text_to_num = {"one":1, "two":2, "three":3, "four":4, "five":5}
    cleaned_data = []    # declare empty List syntax : listName = []
    unique_user = set()  # declare empty set syntax : setName = set()

    for user in  data:
        #clean rating - data inconsistency
        raw_rating = str(user["rating"]).strip().lower()       # strip() : is similar to trim() in java used to truncate left or right extra residue Whitespace
        if(raw_rating in text_to_num):
            raw_rating = text_to_num[raw_rating]
        user["rating"] = raw_rating    # add value in list or if key not exist then add new key with value syntax : distName[key] = val

        # handle missing value
        raw_age = user.get("age")
        if(raw_age == None):
            user["age"] = None

        # deduplication
        if(user["name"].strip() in unique_user):
            continue
        unique_user.add(user["name"])  # add() is used to add any value in set
        cleaned_data.append(user)      # append is used to add value at the end of the list

    return cleaned_data
    
```

```python
cleaned_data = clean_data(data)
print(cleaned_data)
```

```python
def get_insight(cleaned_data):
    avg = 0.0
    count = 0
    for user in cleaned_data:
        avg += float(user["rating"])  # here int(3.5) ❌ gives valueError 
        if(float(user["rating"]) > 3):
            count += 1
    print(f"Average rating: {avg / len(cleaned_data)}")
    print(f"Ratings percentAge above 3: {(count/len(cleaned_data))*100}%")
```

```python
get_insight(cleaned_data)
```
