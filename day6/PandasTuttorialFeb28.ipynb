```python
import pandas as pd
import numpy as np
```

```python
df1 = pd.DataFrame({
    "id" : [1,2,3],
    "name" : ["Mohit","Rohit","Sohit"]
})
df1

df2 = pd.DataFrame({
    "id" : [4,5,6],
    "name" : ["Bablu","Jon","Bob"]
})
df2
#use of Concatenation operator either stack vertically(maintain lables) or horizontally
pd.concat([df1,df2])  # stack vertically == axis = 0
pd.concat([df1,df2],ignore_index = True) # label maintain serially autoincrement type
#stack horizontally
pd.concat([df1,df2],axis=1)
```
