```python
import pandas as pd
import numpy as np
```

```python
# data = pd.read_csv('fortune500.csv', encoding='latin1')
companys = pd.read_csv('Fortune 500 Companies.csv', encoding='ISO-8859-1')
```

```python
companys.shape
```

```python
companys.columns
```

```python
sectors = companys.groupby('sector')
```

```python
# sort all sectors in decreasing order 
sectors.size().sort_values(ascending=False)
```

```python
# Print 1st company of every sector
sectors.first()
```

```python
# print last company of every sector
sectors.last()
```

```python
# group method create a dictionay where key is sector/categoricalCol and value is list of index of all particular sector
sectors.groups  # says index no 9466 -> is belong to category Aerospace & Defense 
companys.iloc[9466,:]  # printing row no 9466
```

```python
# get_group return groups where category = Apparel
sectors.get_group('Apparel')
```
