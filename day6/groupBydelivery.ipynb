```python
import pandas as pd
import numpy as np
```

```python
deliveries = pd.read_csv('deliveries.csv')
```

```python
deliveries.shape
```

```python
runs = deliveries.groupby('batsman')
```

```python
runs.head()
```

```python
# top 5 scorer in the ipl history
runs['batsman_runs'].sum().sort_values(ascending = False).head()
```

```python
# find out top 5 batsman that hit more sixes
mask = deliveries['batsman_runs']==6
new_delivery = deliveries[mask]
new_delivery.groupby('batsman')['batsman_runs'].count().sort_values(ascending = False).head()
```

```python
# top 3 team againgst which virat kohli scored more runs
mask = deliveries['batsman'] == 'V Kohli'
viratPerformance = deliveries[mask]

viratPerformance.groupby('bowling_team')['batsman_runs'] \
                .sum() \
                .sort_values(ascending=False) \
                .head(3)
```

```python
# now create a function that take batsman name and return most run scored against team
def run_scored(batsman):
    mask = deliveries['batsman'] == 'V Kohli'
    viratPerformance = deliveries[mask]

    return viratPerformance.groupby('bowling_team')['batsman_runs'] \
                .sum() \
                .sort_values(ascending=False) \
                .head(1)

    
```

```python
run_scored('DA Warner')
```

```python
# isin() function
# find out most destructive death over batsman in the history of ipl
mask = (deliveries['over'] >= 16) & (deliveries['over'] < 20)
deliveries[mask].groupby('batsman')['batsman_runs'].sum().sort_values(ascending = False).head()
```

```python
death_Over = deliveries[deliveries['over']>15]
all_batter = death_Over.groupby('batsman')['batsman_runs'].count()
x = all_batter>200
batsman_list = all_batter[x].index.tolist()
len(batsman_list)
```

```python
# runs scored by these 43 batsman
# balles played by this 43 batsman
final = deliveries[deliveries['batsman'].isin(batsman_list)]
```

```python
runs = final.groupby('batsman')['batsman_runs'].sum()
balls = final.groupby('batsman')['ball'].count()
balls
sr = (runs/balls)*100
sr.sort_values(ascending=False)
```
