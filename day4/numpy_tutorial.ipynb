```python
#Python List VS numpy Array
'''
Numpy array are faster (optimization in C)
numpy store data in contiguous location   like (1004,1008,1012... )
uses less memory (optimized storage)
built-in mathematical function
supports direct vectorized operations

! why numpy are so fast:
- homogeneous data type and contiguous location
-vector operation and C optimisation
- reduce overhead
'''
size = 1_000_000
py_List = list(range(size))
start = time.time()  # .time() method is similar to Date.now() in js or in java
sq_list = [x*x for x in py_List]
end = time.time()
print(f"Python list takes time = {end-start:.5f} seconds")
print(f"space occupied by python list = {sys.getsizeof(py_List) * len(py_List)} bytes") 

# numpy
npArr = np.array(py_List)
start = time.time()
npsq_list = npArr**2   # direct method for mathematical calculation make it faster than python_list
end = time.time()
print(f"Python list takes time = {end-start:.5f} seconds")
print(f"space occupied by python list = {npArr.nbytes} bytes")  # np_Array.nbytes -> tells size/memory occupied in bytes 

```

```python
#Operations on Arrays
# reshape : means transforming array from dim to another care should be taken size of oldArr and reshaped array always equal
reshapedArr = np_Array.reshape((4,2)) # old dim(2*4) -> new dim (4*2)
print(reshapedArr)
# flatten : means transforming array from n-dim to 1D array
flatten = reshapedArr.flatten()
print(flatten,flatten.shape)
```

```python
# slicing very similar to list syntax : np_Array[start:end:step]
print(flatten[2])
print(flatten[2:]) # [3 4 5 6 7 8]
print(flatten[2:5])
print(flatten[2:-1])
print(flatten[::-1]) #[8 7 6 5 4 3 2 1]
```

```python
# copy vs view
 # python list me copy bnta h sliced part of list ka -> so sliced copy me kiya gya changes ogList me reflect nhi hota but extra memory space aded

 # but in numpy array me view create hota h sliced part ka same array me he so memory efficient but any changes perform on sliced array reflect in ogArray due to view and vice versa
#copy
list = [1,2,3,4,5,6]
slicedList = list[2:-1]  #  # No shared memory with list
print(list,slicedList)
slicedList[0] = -1
print(list,slicedList)  # [1, 2, 3, 4, 5, 6] [-1, 4, 5]

# view
arr = np.array([1,2,3,4,5,6])
slicedArr = arr[2:-1]  # shared memory with arr / shallow copy
print(arr,slicedArr)
slicedArr[0] = -1
print(arr,slicedList)  # [ 1  2 -1  4  5  6] [-1, 4, 5]
#copy method to creat deep copy of npArray
arr1 = np.arange(1,7)
deepCopy = arr1[2:-1].copy() # No shared memory with arr1
deepCopy[0] = -1
print(arr1,deepCopy) # [1 2 3 4 5 6] [-1  4  5]



```

```python
# Broadcasting
# When operating on two arrays, NumPy compares their shapes element-wise. It starts with the trailing (i.e. rightmost) dimension and works its way left. 
# Two dimensions are compatible when

# they are equal, or

# one of them is 1.

# If these conditions are not met, a ValueError: operands could not be broadcast together exception is thrown, indicating that the arrays have incompatible shapes.

a = np.array([1.0, 2.0, 3.0])
b = 2.0
a * b  # scalar b become strech and match the size of a(1,3) ,then via viectorization we can perform arithmetic or mathematical operation on numpy arrays
# array([2.,  4.,  6.])

#example2
import numpy as np
a = np.array([[ 0.0,  0.0,  0.0],
              [10.0, 10.0, 10.0],
              [20.0, 20.0, 20.0],
              [30.0, 30.0, 30.0]])
b = np.array([1.0, 2.0, 3.0])
a + b
# array([[ 1.,  2.,  3.],
#        [11., 12., 13.],
#        [21., 22., 23.],
#        [31., 32., 33.]])
```

```python
# Normalization on vector --> uses concept of BrooadcCasting
# def: normalizing an array means transforming its value so that fit into specific scale/range
# normalization -> says Standard Normalizatio or Z-Score mean
arr = np.array([[1,2],[3,4]])
mean = np.mean(arr)
print(f"mean of array is :{mean}")
# std formula = sqRoot(((val[i] - mean)^2) / N)}
std_dev = np.std(arr)
print(f"Standard deviation of an array is :{std_dev}")
normalized_arr = (arr - mean)/std_dev
print(normalized_arr)
```
