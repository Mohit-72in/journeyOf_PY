```python
# Now i am going to fetch rows based on conditions
# find all matches that that venue = "Mumbai"
def get_count_of_Match_inCity(city): 
    mask = data['city'] == city
    df = data[mask]
    return df.shape[0]
get_count_of_Match_inCity('Mumbai')
```

```python
# find all matches count that were played in chennai after 2010
def get_city_Count_after_Date(city,date):
    mask1 = data['city'] == city
    mask2 = data['date'] > date
    return data[mask1 & mask2].shape[0]
print(get_city_Count_after_Date('Chennai','2018-01-01'))
print(data.shape[0])
data[mask1 & mask2]
```

```python
# value_counts() -> works on series only and worked on categorical_data perfectly but not on numerical data
#      finf out ,frequency count of each unique category 
# Q: find out how many matches each team won along with its win count
data['winner'].value_counts()
```

```python
import matplotlib.pyplot as plt    # used to visualize data using graphs - bar,barh,hist,pie
```

```python
# draw a bar graph where x-asis show categoryname and freq is determined by y-axis
data['winner'].value_counts().plot(kind = 'bar')
```

```python
# draw a vertical bargraph for winner name on y-axis and win count in x-axis
data['winner'].value_counts().plot(kind = 'barh')
```

```python
# toss-decision find using piechart
data['toss_decision'].value_counts().plot(kind = 'pie')
```

```python
# result_margin plot through histo gram where 0 to 20 gives some count which is plotted on histogram
data['result_margin'].value_counts().plot(kind = 'hist')
```

```python
# Some more functions on series
# total match played by 'Mumbai Indians' as team1 or home team
series1 = data['team1'].value_counts()
series2 = data['team2'].value_counts()
series1['Mumbai Indians']   # mumbia as a home team
totalMatchEachTeamPlayed = series1+series2
print(totalMatchEachTeamPlayed['Mumbai Indians'])  # total match played by mumbai as home team or opponent team
```
